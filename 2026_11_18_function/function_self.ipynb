## 第 1 步：定義與呼叫基本函式

使用 `def` 定義函式，名稱後面加上小括號 `()` 與冒號 `:`，內容必須縮排。

```python
def say_hello():
    print("哈囉！歡迎學習 Python 函式。")
    print("函式能幫你把程式碼模組化、重複使用。")

# 呼叫函式
say_hello()
say_hello()
```

## 第 2 步：傳遞參數與引數

支援位置引數（依順序對應）與關鍵字引數（指定參數名稱）。

```python
def greet_user(name, greeting):
    print(f"{greeting}，{name}！")

# 1. 位置引數（依順序對應）
greet_user("小明", "早安")
greet_user("小美", "晚安")

# 2. 關鍵字引數（指定參數名稱）
greet_user(greeting="你好", name="阿強")
```

## 第 3 步：回傳值（return 的重要性）

`return` 能把運算結果交還給呼叫者，並可一次回傳多個值（打包為 Tuple 並解包）。

```python
def add(a, b):
    result = a + b
    return result

total = add(15, 25)
print(f"相加結果：{total}")
print(f"結果兩倍：{total * 2}")

# 回傳多個值
def get_min_and_max(numbers):
    lowest = min(numbers)
    highest = max(numbers)
    return lowest, highest

scores = [85, 92, 58, 77, 99]
min_score, max_score = get_min_and_max(scores)
print(f"最低分：{min_score}，最高分：{max_score}")
```

## 第 4 步：設定預設參數值（Default Arguments）

預設參數必須放在無預設參數後面；切勿使用可變物件（如 `[]`、`{}`）作為預設值。

```python
def power(base, exponent=2):
    return base ** exponent

print(f"5 的平方（預設次方）：{power(5)}")
print(f"5 的三次方：{power(5, 3)}")
print(f"2 的四次方：{power(2, exponent=4)}")

# 安全的預設值寫法
def add_item(item, item_list=None):
    if item_list is None:
        item_list = []
    item_list.append(item)
    return item_list

print(add_item("蘋果"))
print(add_item("香蕉"))
```

## 第 5 步：變數作用域（全域 vs. 區域）

函式內部建立的變數為區域變數，外部無法直接存取。

```python
school_name = "Python 學院"  # 全域變數

def show_student_info(name):
    student_id = "S1001"    # 區域變數
    print(f"學校：{school_name} | 姓名：{name} | 學號：{student_id}")

show_student_info("小華")
```

## 第 6 步：彈性參數（*args 與 **kwargs）與 Docstring

`*args` 接收不定數量位置引數（打包成 Tuple），`**kwargs` 接收關鍵字引數（打包成 Dict）。

```python
def make_order(table_num, *dishes, **details):
    """
    記錄餐廳點餐資訊。
    :param table_num: 桌號
    :param dishes: 不定數量的餐點名稱（tuple）
    :param details: 不定數量的備註資訊（dict）
    """
    print(f"=== 桌號 {table_num} 的點餐 ===")
    print(f"餐點項目：{dishes}")
    print(f"特殊備註：{details}")

make_order(5, "牛肉麵", "燙青菜", "滷蛋", spicy="少辣", soup="多湯")
```

## 練習題

1. 溫度轉換函式 `celsius_to_fahrenheit(c)`
2. 統計計算函式 `calculate_stats(numbers)`
3. 個人檔案產生器 `create_profile(name, age, city="台北")`
4. 不定量加總函式 `sum_all(*numbers)`

```python
# 練習 1：溫度轉換
def celsius_to_fahrenheit(c):
    return c * 9 / 5 + 32

print("[練習 1] 攝氏 25 度轉華氏:", celsius_to_fahrenheit(25))

# 練習 2：統計計算
def calculate_stats(numbers):
    total = sum(numbers)
    avg = total / len(numbers)
    highest = max(numbers)
    lowest = min(numbers)
    return total, avg, highest, lowest

t, a, h, l = calculate_stats([10, 20, 30, 40, 50])
print(f"[練習 2] 總和={t}, 平均={a}, 最大={h}, 最小={l}")

# 練習 3：個人檔案產生器
def create_profile(name, age, city="台北"):
    return {"name": name, "age": age, "city": city}

print("[練習 3 預設]", create_profile("王小明", 20))
print("[練習 3 指定]", create_profile("李小華", 22, city="高雄"))

# 練習 4：不定量加總
def sum_all(*numbers):
    return sum(numbers)

print("[練習 4 無參數] sum_all():", sum_all())
print("[練習 4 多參數] sum_all(1, 2, 3, 4, 5):", sum_all(1, 2, 3, 4, 5))
```
